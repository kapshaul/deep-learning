# XOR Classification with a Hidden Layer

XOR returns 1 when its two binary inputs differ and 0 when they match. The four input/target pairs are `(0, 0) → 0`, `(0, 1) → 1`, `(1, 0) → 1`, and `(1, 1) → 0`. A single linear decision boundary cannot separate the two classes.

A hidden layer with nonlinear activations lets the model combine multiple boundaries. One useful logical interpretation is:

```text
XOR(x1, x2) = (x1 OR x2) AND NOT(x1 AND x2)
```

The network learns continuous weights and biases rather than assigning fixed Boolean gates to its hidden neurons.

## Data and execution

Run the cells from top to bottom. The data cell repeats the four truth-table rows 5,000 times, producing 20,000 training rows but only four distinct inputs. The reported accuracy is training accuracy; no separate evaluation split is defined. See the [README](README.md#xor-classification) for installation and a four-case prediction check.


````
import tensorflow as tf
import numpy as np
````

````
# XOR input output data
x = [
     [0, 0],
     [0, 1],
     [1, 0],
     [1, 1]
] * 5000

y = [
     [0],
     [1],
     [1],
     [0]
] * 5000

x = np.array(x)
y = np.array(y)
````

````
def print_weights(model, state):
    print()
    print(f"Model weights {state}:")
    weights = model.get_weights()[0]
    for i in range(len(weights)):
        print(f"{i} layer weights = {weights[i]}")
    print()
````

## Model and training

The model takes two scalar inputs, uses two sigmoid hidden units, and produces one sigmoid output:

```python
model = tf.keras.Sequential([
    tf.keras.layers.Dense(2, activation="sigmoid", input_shape=(2,)),
    tf.keras.layers.Dense(1, activation="sigmoid")
])
```

The next cell trains with binary cross-entropy and Adam (learning rate `0.1`) for 10 epochs with batches of 64. Initialization is not seeded, so convergence can vary between runs. The `print_weights` helper shows only the first layer's weight matrix; its row labels do not refer to different network layers.


````
# Define the neural network model
model = tf.keras.Sequential([
    # Fully connected layer with sigmoid activation
    tf.keras.layers.Dense(2, activation="sigmoid", input_shape=(2,)),

    # Output layer with sigmoid activation
    tf.keras.layers.Dense(1, activation="sigmoid")
])

model.summary()

# Model weights before training
print_weights(model, "before")

optimizer = tf.keras.optimizers.Adam(learning_rate=0.1)
model.compile(optimizer=optimizer, loss="binary_crossentropy", metrics=['accuracy'])

model.fit(x, y, batch_size=64, epochs=10)

# Model weights after training
print_weights(model, "after")
````
